```python
import ast
import pandas as pd
import numpy as np
```

```python
def convert(text):
    try:
        return ast.literal_eval(text) if isinstance(text, str) else []
    except (ValueError, SyntaxError):
        return []
```

```python
def extract_names(obj):
    return [
        item["name"]
        for item in convert(obj)
        if isinstance(item, dict) and item.get("name")
    ]

movies_merged["genres_list"] = movies_merged["genres"].apply(extract_names)
movies_merged["keywords_list"] = movies_merged["keywords"].apply(extract_names)

display(
    movies_merged[
        ["title_movie", "genres_list", "keywords_list"]
    ].head(5)
)
```

```python
def extract_cast(obj):
    return [
        item["name"]
        for item in convert(obj)[:3]
        if isinstance(item, dict) and item.get("name")
    ]

movies_merged["cast_list"] = movies_merged["cast"].apply(extract_cast)

display(
    movies_merged[["title_movie", "cast_list"]].head(5)
)
```

```python
def extract_director(obj):
    return [
        item["name"]
        for item in convert(obj)
        if isinstance(item, dict)
        and item.get("job") == "Director"
        and item.get("name")
    ][:1]

movies_merged["director_list"] = movies_merged["crew"].apply(extract_director)

display(
    movies_merged[["title_movie", "director_list"]].head(5)
)
```

```python
# Keep only the useful features for recommendations
recommendation_data = movies_merged[
    [
        "id",
        "title_movie",
        "genres_list",
        "keywords_list",
        "cast_list",
        "director_list",
        "overview"
    ]
].copy()

# Replace missing overviews
recommendation_data["overview"] = (
    recommendation_data["overview"].fillna("")
)

# Convert text to lowercase and remove extra spaces
recommendation_data["overview"] = (
    recommendation_data["overview"]
    .astype(str)
    .str.lower()
    .str.strip()
)

print("Recommendation dataset shape:", recommendation_data.shape)
display(recommendation_data.head())
```

```python
def combine_features(row):
    features = (
        row["genres_list"]
        + row["keywords_list"]
        + row["cast_list"]
        + row["director_list"]
    )

    # Normalize names and words
    features = [
        str(item).lower().replace(" ", "")
        for item in features
        if item
    ]

    overview = row["overview"]
    return " ".join(features) + " " + overview


recommendation_data["combined_features"] = (
    recommendation_data.apply(combine_features, axis=1)
)

display(
    recommendation_data[
        ["title_movie", "combined_features"]
    ].head(5)
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("TF-IDF and cosine similarity imported successfully!")
```

```python
# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

# Convert combined movie features into vectors
tfidf_matrix = tfidf.fit_transform(
    recommendation_data["combined_features"].fillna("")
)

print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Number of movies:", len(recommendation_data))
print("Number of text features:", tfidf_matrix.shape[1])
```

```python
# Calculate cosine similarity between movie vectors
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
# Check similarity of the first movie with itself
print(
    "Similarity of the first movie with itself:",
    similarity_matrix[0, 0]
)

# Display similarity scores for the first movie
print("\nFirst five similarity scores:")
print(similarity_matrix[0][:5])
```

```python
# Create a mapping between movie titles and their indices
recommendation_data = recommendation_data.reset_index(drop=True)

indices = pd.Series(
    recommendation_data.index,
    index=recommendation_data["title_movie"].str.lower()
).drop_duplicates()


def recommend_movies(movie_title, top_n=5):
    movie_title = movie_title.strip().lower()

    if movie_title not in indices:
        print(f"Movie '{movie_title}' not found in the dataset.")
        return pd.DataFrame()

    idx = indices[movie_title]

    # Get similarity scores for the selected movie
    scores = list(enumerate(similarity_matrix[idx]))

    # Sort by similarity, excluding the selected movie itself
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    scores = [item for item in scores if item[0] != idx][:top_n]

    # Return recommended movie titles and similarity scores
    results = recommendation_data.iloc[[i[0] for i in scores]][
        ["title_movie"]
    ].copy()

    results["similarity_score"] = [round(i[1], 4) for i in scores]
    results = results.rename(columns={"title_movie": "recommended_movie"})

    return results.reset_index(drop=True)


print("Recommendation function created successfully!")
```

```python
def evaluate_recommendations(movie_titles, top_n=5):
    results = []

    for title in movie_titles:
        recommendations = recommend_movies(title, top_n=top_n)

        results.append({
            "input_movie": title,
            "recommendations_returned": len(recommendations),
            "no_self_recommendation": (
                title.lower() not in
                recommendations["recommended_movie"].str.lower().tolist()
            ) if not recommendations.empty else False,
            "unique_recommendations": (
                recommendations["recommended_movie"].is_unique
            ) if not recommendations.empty else False
        })

    return pd.DataFrame(results)


evaluation_results = evaluate_recommendations(test_movies)
display(evaluation_results)
```

```python
import json
from pathlib import Path

# Create output folder
output_dir = Path("web_data")
output_dir.mkdir(exist_ok=True)

# Prepare movie lookup
movie_titles = recommendation_data["title_movie"].tolist()

# Generate top 10 recommendations per movie
recommendation_dict = {}

for idx, title in enumerate(movie_titles):
    scores = list(enumerate(similarity_matrix[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)

    # Exclude the movie itself
    scores = [
        (movie_idx, score)
        for movie_idx, score in scores
        if movie_idx != idx
    ][:10]

    recommendation_dict[title] = [
        {
            "title": movie_titles[movie_idx],
            "score": round(float(score), 4)
        }
        for movie_idx, score in scores
    ]

# Save JSON
output_path = output_dir / "recommendations.json"

with open(output_path, "w", encoding="utf-8") as file:
    json.dump(recommendation_dict, file, ensure_ascii=False)

print("Recommendations exported successfully!")
print("Total movies:", len(recommendation_dict))
print("Saved to:", output_path)
print("File size (MB):", round(output_path.stat().st_size / (1024 * 1024), 2))
```

```python
import pandas as pd
import json
from pathlib import Path

BASE_DIR = Path.cwd()

movies = pd.read_csv(BASE_DIR / "data" / "tmdb_5000_movies.csv")
credits = pd.read_csv(BASE_DIR / "data" / "tmdb_5000_credits.csv")

print("Movies shape:", movies.shape)
print("Credits shape:", credits.shape)

display(movies.head(2))
```

```python
import ast

movies_clean = movies[
    ["id", "title", "genres", "keywords", "overview"]
].copy()

movies_clean["overview"] = movies_clean["overview"].fillna("")

movies_clean = movies_clean.dropna(subset=["id", "title"])

movies_merged = movies_clean.merge(
    credits,
    left_on="id",
    right_on="movie_id",
    how="inner",
    suffixes=("_movie", "_credit")
)

print("Merged dataset:", movies_merged.shape)
print(movies_merged.columns.tolist())
```

```python

import json
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd()

# Load the existing exported movie data to preserve poster paths.
movies_json_path = BASE_DIR / "web_data" / "movies.json"

with open(movies_json_path, "r", encoding="utf-8") as f:
    old_movies = json.load(f)

old_by_id = {str(movie["id"]): movie for movie in old_movies}

# Map the original movie metadata by movie ID.
metadata = movies.set_index("id")

frontend_movies = []

for _, row in movies_merged.iterrows():
    movie_id = int(row["id"])
    original = metadata.loc[movie_id]
    previous = old_by_id.get(str(movie_id), {})

    release_date = original.get("release_date", "")
    if pd.isna(release_date):
        release_date = ""
    else:
        release_date = str(release_date)

    vote_average = original.get("vote_average", 0)
    if pd.isna(vote_average):
        vote_average = 0.0
    else:
        vote_average = float(vote_average)

    genres = row.get("genres", "[]")
    try:
        genres = json.loads(genres) if isinstance(genres, str) else []
        genres = [
            genre.get("name", "")
            for genre in genres
            if isinstance(genre, dict) and genre.get("name")
        ]
    except (ValueError, TypeError):
        genres = []

    overview = row.get("overview", "")
    if pd.isna(overview):
        overview = ""

    frontend_movies.append({
        "id": movie_id,
        "title": str(row["title_movie"]),
        "overview": str(overview),
        "genres": genres,
        "release_date": release_date,
        "vote_average": vote_average,
        "poster_path": previous.get("poster_path")
    })

with open(movies_json_path, "w", encoding="utf-8") as f:
    json.dump(frontend_movies, f, ensure_ascii=False)

print("Export completed:", len(frontend_movies), "movies")
print(
    "Movies with posters:",
    sum(bool(movie["poster_path"]) for movie in frontend_movies)
)
print(
    "Movies with release dates:",
    sum(bool(movie["release_date"]) for movie in frontend_movies)
)
print("Sample movie:", frontend_movies[0])
```
