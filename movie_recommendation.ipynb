```python
# Select the columns needed for recommendations
movies_clean = movies[
    ["id", "title", "genres", "keywords", "overview"]
].copy()

# Handle missing movie descriptions
movies_clean["overview"] = movies_clean["overview"].fillna("")

# Remove rows with missing IDs or titles
movies_clean = movies_clean.dropna(subset=["id", "title"])

# Display the results
print("Original movies shape:", movies.shape)
print("Cleaned movies shape:", movies_clean.shape)

display(movies_clean.head())
```

```python
# Merge movie details with cast and crew information
movies_merged = movies_clean.merge(
    credits,
    left_on="id",
    right_on="movie_id",
    how="inner",
    suffixes=("_movie", "_credit")
)

print("Merged dataset shape:", movies_merged.shape)
print("\nColumns:")
print(movies_merged.columns.tolist())

display(movies_merged.head(3))
```

```python
print("Number of movies:", len(movies_merged))
print("Missing overviews:", movies_merged["overview"].isnull().sum())
print("Missing cast entries:", movies_merged["cast"].isnull().sum())
print("Missing crew entries:", movies_merged["crew"].isnull().sum())

display(
    movies_merged[
        ["title_movie", "genres", "keywords", "overview", "cast", "crew"]
    ].head(5)
)
```

```python
import ast
import pandas as pd
import numpy as np
```

```python
def convert(text):
    try:
        return ast.literal_eval(text) if isinstance(text, str) else []
    except (ValueError, SyntaxError):
        return []
```

```python
def extract_names(obj):
    return [
        item["name"]
        for item in convert(obj)
        if isinstance(item, dict) and item.get("name")
    ]

movies_merged["genres_list"] = movies_merged["genres"].apply(extract_names)
movies_merged["keywords_list"] = movies_merged["keywords"].apply(extract_names)

display(
    movies_merged[
        ["title_movie", "genres_list", "keywords_list"]
    ].head(5)
)
```

```python
def extract_cast(obj):
    return [
        item["name"]
        for item in convert(obj)[:3]
        if isinstance(item, dict) and item.get("name")
    ]

movies_merged["cast_list"] = movies_merged["cast"].apply(extract_cast)

display(
    movies_merged[["title_movie", "cast_list"]].head(5)
)
```

```python
def extract_director(obj):
    return [
        item["name"]
        for item in convert(obj)
        if isinstance(item, dict)
        and item.get("job") == "Director"
        and item.get("name")
    ][:1]

movies_merged["director_list"] = movies_merged["crew"].apply(extract_director)

display(
    movies_merged[["title_movie", "director_list"]].head(5)
)
```

```python
# Keep only the useful features for recommendations
recommendation_data = movies_merged[
    [
        "id",
        "title_movie",
        "genres_list",
        "keywords_list",
        "cast_list",
        "director_list",
        "overview"
    ]
].copy()

# Replace missing overviews
recommendation_data["overview"] = (
    recommendation_data["overview"].fillna("")
)

# Convert text to lowercase and remove extra spaces
recommendation_data["overview"] = (
    recommendation_data["overview"]
    .astype(str)
    .str.lower()
    .str.strip()
)

print("Recommendation dataset shape:", recommendation_data.shape)
display(recommendation_data.head())
```

```python
def combine_features(row):
    features = (
        row["genres_list"]
        + row["keywords_list"]
        + row["cast_list"]
        + row["director_list"]
    )

    # Normalize names and words
    features = [
        str(item).lower().replace(" ", "")
        for item in features
        if item
    ]

    overview = row["overview"]
    return " ".join(features) + " " + overview


recommendation_data["combined_features"] = (
    recommendation_data.apply(combine_features, axis=1)
)

display(
    recommendation_data[
        ["title_movie", "combined_features"]
    ].head(5)
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

print("TF-IDF and cosine similarity imported successfully!")
```

```python
# Create the TF-IDF vectorizer
tfidf = TfidfVectorizer(
    max_features=5000,
    stop_words="english"
)

# Convert combined movie features into vectors
tfidf_matrix = tfidf.fit_transform(
    recommendation_data["combined_features"].fillna("")
)

print("TF-IDF matrix shape:", tfidf_matrix.shape)
print("Number of movies:", len(recommendation_data))
print("Number of text features:", tfidf_matrix.shape[1])
```

```python
# Calculate cosine similarity between movie vectors
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
# Check similarity of the first movie with itself
print(
    "Similarity of the first movie with itself:",
    similarity_matrix[0, 0]
)

# Display similarity scores for the first movie
print("\nFirst five similarity scores:")
print(similarity_matrix[0][:5])
```

```python
# Create a mapping between movie titles and their indices
recommendation_data = recommendation_data.reset_index(drop=True)

indices = pd.Series(
    recommendation_data.index,
    index=recommendation_data["title_movie"].str.lower()
).drop_duplicates()


def recommend_movies(movie_title, top_n=5):
    movie_title = movie_title.strip().lower()

    if movie_title not in indices:
        print(f"Movie '{movie_title}' not found in the dataset.")
        return pd.DataFrame()

    idx = indices[movie_title]

    # Get similarity scores for the selected movie
    scores = list(enumerate(similarity_matrix[idx]))

    # Sort by similarity, excluding the selected movie itself
    scores = sorted(scores, key=lambda x: x[1], reverse=True)
    scores = [item for item in scores if item[0] != idx][:top_n]

    # Return recommended movie titles and similarity scores
    results = recommendation_data.iloc[[i[0] for i in scores]][
        ["title_movie"]
    ].copy()

    results["similarity_score"] = [round(i[1], 4) for i in scores]
    results = results.rename(columns={"title_movie": "recommended_movie"})

    return results.reset_index(drop=True)


print("Recommendation function created successfully!")
```

```python
# Check whether the selected movie appears in its own recommendations
movie_title = "The Dark Knight"

results = recommend_movies(movie_title, top_n=5)

if not results.empty:
    recommended_titles = results["recommended_movie"].str.lower().tolist()

    print("Number of recommendations:", len(results))
    print(
        "Selected movie excluded:",
        movie_title.lower() not in recommended_titles
    )
    print(
        "Recommendations are unique:",
        results["recommended_movie"].is_unique
    )
    display(results)
```

```python
test_movies = [
    "The Dark Knight",
    "Avatar",
    "Iron Man",
    "Titanic",
    "The Avengers"
]

for movie in test_movies:
    print(f"\nRecommendations for: {movie}")
    display(recommend_movies(movie, top_n=5))
```

```python
def evaluate_recommendations(movie_titles, top_n=5):
    results = []

    for title in movie_titles:
        recommendations = recommend_movies(title, top_n=top_n)

        results.append({
            "input_movie": title,
            "recommendations_returned": len(recommendations),
            "no_self_recommendation": (
                title.lower() not in
                recommendations["recommended_movie"].str.lower().tolist()
            ) if not recommendations.empty else False,
            "unique_recommendations": (
                recommendations["recommended_movie"].is_unique
            ) if not recommendations.empty else False
        })

    return pd.DataFrame(results)


evaluation_results = evaluate_recommendations(test_movies)
display(evaluation_results)
```
